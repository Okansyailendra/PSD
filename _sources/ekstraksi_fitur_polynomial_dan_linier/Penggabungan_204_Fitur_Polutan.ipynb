# Penggabungan Fitur 3 Polutan (204 Kolom)
## Urutan: NO2, SO2, CO (Linier vs Polinomial)

Notebook ini menggabungkan hasil ekstraksi 68 fitur TSFEL dari 3 jenis polutan menjadi satu dataset terpadu berukuran **204 kolom**, dengan urutan baku:
1. **NO2** (Kolom 1 – 68)
2. **SO2** (Kolom 69 – 136)
3. **CO**  (Kolom 137 – 204)

### Rincian Kolom:
- NO2 : 68 fitur (prefix `NO2_`)
- SO2 : 68 fitur (prefix `SO2_`)
- CO  : 68 fitur (prefix `CO_`)
- **Total**: $68 + 68 + 68 = 204$ kolom numerik.


### Bagian 1: Penggabungan 204 Kolom — Versi Polinomial
Urutan penggabungan horizontal: `NO2` -> `SO2` -> `CO`.


```python
import pandas as pd

# 1. Membaca masing-masing file CSV polutan versi Polinomial (urutan: NO2, SO2, CO)
df_no2_poly = pd.read_csv('03_Ekstraksi_Fitur_Polinomial/NO2_Manyar_TSFEL_Polinomial.csv')
df_so2_poly = pd.read_csv('03_Ekstraksi_Fitur_Polinomial/SO2_Manyar_TSFEL_Polinomial.csv')
df_co_poly = pd.read_csv('03_Ekstraksi_Fitur_Polinomial/CO_Manyar_TSFEL_Polinomial.csv')

print('Ukuran masing-masing file polinomial:')
print(f'  NO2 : {df_no2_poly.shape}')
print(f'  SO2 : {df_so2_poly.shape}')
print(f'  CO  : {df_co_poly.shape}')

# 2. Menambahkan prefix polutan agar nama kolom terstruktur dan unik
df_no2_poly = df_no2_poly.add_prefix('NO2_')
df_so2_poly = df_so2_poly.add_prefix('SO2_')
df_co_poly = df_co_poly.add_prefix('CO_')

# 3. Menggabungkan ketiga polutan sesuai urutan: NO2, SO2, CO
df_204_poly = pd.concat([df_no2_poly, df_so2_poly, df_co_poly], axis=1)
print(f'Hasil penggabungan Polinomial (NO2, SO2, CO): {df_204_poly.shape}')

# 4. Menyimpan ke file CSV baru
output_poly = '204 kolom polynomial.csv'
df_204_poly.to_csv(output_poly, index=False)
print(f'File berhasil disimpan ke: {output_poly}')
```

### Bagian 2: Penggabungan 204 Kolom — Versi Linier
Urutan penggabungan horizontal: `NO2` -> `SO2` -> `CO`.


```python
# 1. Membaca masing-masing file CSV polutan versi Linier (urutan: NO2, SO2, CO)
df_no2_lin = pd.read_csv('02_Ekstraksi_Fitur_Linear/NO2_Manyar_TSFEL.csv')
df_so2_lin = pd.read_csv('02_Ekstraksi_Fitur_Linear/SO2_Manyar_TSFEL.csv')
df_co_lin = pd.read_csv('02_Ekstraksi_Fitur_Linear/CO_Manyar_TSFEL.csv')

print('Ukuran masing-masing file linier:')
print(f'  NO2 : {df_no2_lin.shape}')
print(f'  SO2 : {df_so2_lin.shape}')
print(f'  CO  : {df_co_lin.shape}')

# 2. Menambahkan prefix nama polutan
df_no2_lin = df_no2_lin.add_prefix('NO2_')
df_so2_lin = df_so2_lin.add_prefix('SO2_')
df_co_lin = df_co_lin.add_prefix('CO_')

# 3. Menggabungkan ketiga polutan sesuai urutan: NO2, SO2, CO
df_204_lin = pd.concat([df_no2_lin, df_so2_lin, df_co_lin], axis=1)
print(f'Hasil penggabungan Linier (NO2, SO2, CO): {df_204_lin.shape}')

# 4. Menyimpan ke file CSV baru
output_lin = '204 kolom linier.csv'
df_204_lin.to_csv(output_lin, index=False)
print(f'File berhasil disimpan ke: {output_lin}')
```

### Bagian 3: Verifikasi Urutan dan Kualitas Kolom
Memeriksa urutan blok kolom (NO2, SO2, CO) serta memastikan tidak ada *missing value* (`NaN`) atau nilai tak hingga (`Inf`).


```python
import numpy as np

print('=== Verifikasi Urutan Blok Kolom ===')
print(f"Kolom 1 - 5   (Blok NO2): {df_204_poly.columns[0:5].tolist()}")
print(f"Kolom 69 - 73 (Blok SO2): {df_204_poly.columns[68:73].tolist()}")
print(f"Kolom 137-141 (Blok CO) : {df_204_poly.columns[136:141].tolist()}")

print('\n=== Ringkasan Kualitas Data ===')
print(f"[Polinomial] Total Kolom: {df_204_poly.shape[1]} | Missing NaN: {df_204_poly.isna().sum().sum()} | Nilai Inf: {np.isinf(df_204_poly.values).sum()}")
print(f"[Linier]     Total Kolom: {df_204_lin.shape[1]} | Missing NaN: {df_204_lin.isna().sum().sum()} | Nilai Inf: {np.isinf(df_204_lin.values).sum()}")
```
